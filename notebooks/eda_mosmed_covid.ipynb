```python
%matplotlib widget

from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display
from tqdm import tqdm

import nibabel as nib
import numpy as np
import pandas as pd


import cv2
```

```python
d_path = "../data/uncompressed/MosMedData Chest CT Scans with COVID-19 Related Findings COVID19_1110 1.0/studies"
d_path = Path(d_path)

ct_0 = d_path / "CT-0"
ct_1 = d_path / "CT-1"
ct_2 = d_path / "CT-2"
ct_3 = d_path / "CT-3"
ct_4 = d_path / "CT-4"

ct_0_files = list(ct_0.glob("*.nii"))
ct_1_files = list(ct_1.glob("*.nii"))
ct_2_files = list(ct_2.glob("*.nii"))
ct_3_files = list(ct_3.glob("*.nii"))
ct_4_files = list(ct_4.glob("*.nii"))
print(f"CT-0 files: {len(ct_0_files)}")
print(f"CT-1 files: {len(ct_1_files)}")
print(f"CT-2 files: {len(ct_2_files)}")
print(f"CT-3 files: {len(ct_3_files)}")
print(f"CT-4 files: {len(ct_4_files)}")

print(f"CT-0 size: {sum(f.stat().st_size for f in ct_0_files) / 1e6:.2f} MB")
print(f"CT-1 size: {sum(f.stat().st_size for f in ct_1_files) / 1e6:.2f} MB")
print(f"CT-2 size: {sum(f.stat().st_size for f in ct_2_files) / 1e6:.2f} MB")
print(f"CT-3 size: {sum(f.stat().st_size for f in ct_3_files) / 1e6:.2f} MB")
print(f"CT-4 size: {sum(f.stat().st_size for f in ct_4_files) / 1e6:.2f} MB")
```

```python
slices = []
sample_size = len(ct_0_files) // 10
for f in tqdm(ct_0_files[:sample_size]):
    data = nib.load(f).get_fdata()
    slices.extend([data[:, :, i] for i in range(data.shape[2])])

ct_0_data = np.stack(slices, axis=0)
print(f"CT-0 data shape: {ct_0_data.shape}")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(10, 5))
ax[0].imshow(ct_0_data[0], cmap="gray")
ax[0].set_title("Slice 0")
ax[0].axis("off")
ax[0].grid(False)

ax[1].imshow(ct_0_data[ct_0_data.shape[0] // 2], cmap="gray")
ax[1].set_title("Slice Middle")
ax[1].axis("off")
ax[1].grid(False)

ax[2].imshow(ct_0_data[-1], cmap="gray")
ax[2].set_title("Slice Last")
ax[2].axis("off")
ax[2].grid(False)

plt.tight_layout()
plt.show()
```

```python
total_slices = [0] * 5
cats = [ct_0_files, ct_1_files, ct_2_files, ct_3_files, ct_4_files]
for cat_idx, cat in enumerate(cats):
    print(f"Processing category {cat_idx} with {len(cat)} files")
    for f in tqdm(cat):
        data = nib.load(f).get_fdata()
        total_slices[cat_idx] += data.shape[2]

slices_df = pd.DataFrame({
    "Category": ["CT-0", "CT-1", "CT-2", "CT-3", "CT-4"],
    "Total Slices": total_slices
})
slices_df
```
